# 🔴 困难：GRPO 损失

实现**分组相对策略优化（Group Relative Policy Optimization，GRPO）**损失——一种常用于 RLAIF（基于 AI 反馈的强化学习）的组内基线减除 REINFORCE 目标。

给定一批对数概率、标量奖励和分组 id（每个提示对应一组），定义组内归一化优势：

$$A_i = \frac{r_i - \bar r_{g(i)}}{\text{std}_{g(i)} + \epsilon}$$

其中 \(\bar r_{g(i)}\) 和 \(\text{std}_{g(i)}\) 是样本 \(i\) 所在组的奖励均值和标准差。

GRPO 损失为负优势加权对数概率：

$$\mathcal{L}_{\text{GRPO}} = -\mathbb{E}_i \big[\,\text{stop\_grad}(A_i)\, \log \pi_\theta(y_i)\big].$$
