# 🔴 困难：多头注意力

从零实现**多头注意力（Multi-Head Attention）**——Transformer 的核心构建块。

$$\text{MultiHead}(Q, K, V) = \text{Concat}(\text{head}_1, \dots, \text{head}_h) W^O$$
$$\text{head}_i = \text{Attention}(Q W_i^Q,\; K W_i^K,\; V W_i^V)$$

### 函数签名
```python
class MultiHeadAttention:
    def __init__(self, d_model: int, num_heads: int): ...
    def forward(self, Q, K, V) -> torch.Tensor: ...
```

### 要求
- 使用 `nn.Linear(d_model, d_model)` 作为 `self.W_q`、`self.W_k`、`self.W_v`、`self.W_o`
- 每个注意力头的 `d_k = d_model // num_heads`
- `forward(Q, K, V)`：Q 形状为 `(B, seq_q, d_model)`，K/V 形状为 `(B, seq_k, d_model)`
- 必须支持**交叉注意力**（`seq_q != seq_k`）
- **不得**使用 `torch.nn.MultiheadAttention`
- **可以**使用 `torch.softmax` 和 `torch.matmul`

### 步骤
1. 投影：`q = self.W_q(Q)`，`k = self.W_k(K)`，`v = self.W_v(V)`
2. 变形为 `(B, num_heads, seq, d_k)`
3. 每个注意力头进行缩放点积注意力
4. 拼接所有头 → `(B, seq_q, d_model)`
5. 输出投影：`self.W_o(concat)`
