# 🔴 困难：分组查询注意力（GQA）

实现**分组查询注意力（Grouped Query Attention）**——LLaMA 2、Mistral 等模型中用于减小 KV 缓存的技术。

与 MHA 类似，但 KV 头数**少于** Q 头数。每组 Q 头共享同一个 K/V 头。

### 函数签名
```python
class GroupQueryAttention:
    def __init__(self, d_model: int, num_heads: int, num_kv_heads: int): ...
    def forward(self, x) -> torch.Tensor:  # 自注意力
```

### 要求
- `self.W_q`：`nn.Linear(d_model, d_model)` — 完整的 Q 投影
- `self.W_k`：`nn.Linear(d_model, num_kv_heads * d_k)` — 缩减的 K 投影
- `self.W_v`：`nn.Linear(d_model, num_kv_heads * d_k)` — 缩减的 V 投影
- `self.W_o`：`nn.Linear(d_model, d_model)` — 输出投影
- `d_k = d_model // num_heads`
- 使用 `repeat_interleave` 扩展 KV 头以匹配 Q 头数
- 当 `num_kv_heads == num_heads` 时，行为应与标准 MHA 相同
