# 🟡 中等：实现 LayerNorm

从零实现**层归一化（Layer Normalization）**。

$$\text{LayerNorm}(x) = \gamma \cdot \frac{x - \mu}{\sqrt{\sigma^2 + \epsilon}} + \beta$$

其中 $\mu$ 和 $\sigma^2$ 在**最后一个维度**上计算。

```python
    eps: float = 1e-5
) -> torch.Tensor:
    ...
```

- **不得**使用 `F.layer_norm` 或 `torch.nn.LayerNorm`
