# 🔴 困难：线性自注意力

实现**线性注意力（Linear Attention）**——时间复杂度为 O(S·D²) 而非 O(S²·D)，可高效处理长序列。

用**核特征映射** $\phi$ 替换 softmax：

$$\text{LinearAttn}(Q,K,V) = \frac{\phi(Q) \left(\phi(K)^T V\right)}{\phi(Q) \cdot \sum \phi(K)}$$

### 特征映射
使用 $\phi(x) = \text{elu}(x) + 1$（确保特征非负）。

### 关键思路
不计算 $S \times S$ 的注意力矩阵，而是先计算 $\phi(K)^T V$（一个 $D_k \times D_v$ 的矩阵），再乘以 $\phi(Q)$。

### 规则
- 必须使用特征映射（**不能**使用 softmax）
- 必须是 O(S·D²) 复杂度——应在长序列上快速运行
- **可以**使用 `F.elu`
