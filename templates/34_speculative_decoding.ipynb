```python
    # target_probs: (K, V) 来自目标（大）模型
    # draft_probs: (K, V) 来自草稿（小）模型
    # draft_tokens: (K,) 由草稿模型采样的 token
    # 返回：已接受的 token 列表（1 到 K 个）
```

### 算法
对每个位置 i = 0, ..., K-1：
1. `ratio = target_probs[i, token_i] / draft_probs[i, token_i]`
2. 以概率 `min(1, ratio)` 接受
3. 如果拒绝：从 `normalize(max(0, target - draft))` 中采样，追加后停止
