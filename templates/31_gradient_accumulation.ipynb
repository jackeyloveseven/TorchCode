### 算法
1. `optimizer.zero_grad()`
2. 对 micro_batches 中每个 `(x, y)`：`loss = loss_fn(model(x), y) / len(micro_batches)`，然后 `loss.backward()`
3. `optimizer.step()`
4. 返回累积的总损失

关键思路：在反向传播前将每个损失除以 `n`，使累积梯度等价于单次大批量梯度。
