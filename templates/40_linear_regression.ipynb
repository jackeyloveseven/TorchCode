# 🟡 中等：线性回归

用三种不同方法实现**线性回归**——全部使用纯 PyTorch。

给定形状为 `(N, D)` 的数据 `X` 和形状为 `(N,)` 的目标 `y`，找到形状为 `(D,)` 的权重 `w` 和标量偏置 `b`，使得：

$$\hat{y} = Xw + b$$

### 函数签名
```python
class LinearRegression:
    def closed_form(self, X: Tensor, y: Tensor) -> tuple[Tensor, Tensor]: ...
    def gradient_descent(self, X: Tensor, y: Tensor, lr=0.01, steps=1000) -> tuple[Tensor, Tensor]: ...
    def nn_linear(self, X: Tensor, y: Tensor, lr=0.01, steps=1000) -> tuple[Tensor, Tensor]: ...
```

所有方法返回 `(w, b)`，其中 `w` 形状为 `(D,)`，`b` 形状为 `()`。

### 方法一——解析解（正规方程）
用全一列扩充 X，然后求解：

$$\theta = (X_{aug}^T X_{aug})^{-1} X_{aug}^T y$$

或使用 `torch.linalg.lstsq` / `torch.linalg.solve`。

### 方法二——手动梯度下降
将 `w` 和 `b` 初始化为零。重复 `steps` 次：
```
pred = X @ w + b
error = pred - y
grad_w = (2/N) * X^T @ error
grad_b = (2/N) * error.sum()
w -= lr * grad_w
b -= lr * grad_b
```

### 方法三——PyTorch nn.Linear
创建 `nn.Linear(D, 1)`，使用 `nn.MSELoss` 和优化器（如 `torch.optim.SGD`）。
训练后从该层提取 `w` 和 `b`。

### 规则
- 所有输入和输出必须是 **PyTorch 张量**
- 不得使用 numpy 或 sklearn
- `closed_form` 不得使用迭代优化
- `gradient_descent` 必须手动计算梯度（不使用 `autograd`）
- `nn_linear` 应使用 `torch.nn.Linear` 和 `loss.backward()`
